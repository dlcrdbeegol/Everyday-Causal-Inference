```python
import pandas as pd
import statsmodels.formula.api as smf
```

```python
git = "https://raw.githubusercontent.com/RobsonTigre/everyday-ci/main/data/"
data = pd.read_csv(git+"advertising_data.csv")
```

```python
# simple regression
simple_regression = smf.ols('sales ~ ad_spend' , data = data).fit()
```

```python
simple_regression.summary()
```

```python
multiple_regression = smf.ols('sales ~ ad_spend + holiday', data=data).fit()
```

```python
print(multiple_regression.summary())
```

```python
df = pd.read_csv(git+'tricky_coefficients.csv')

m1 = smf.ols('time_on_app ~ new_ui', data=df).fit()
print(m1.summary())
```

```python
# We explicitly specify the reference group
m2 = smf.ols('time_on_app ~ new_ui + C(user_segment, Treatment(reference="Casual"))', 
        data=df).fit()
print(m2.summary())
```

```python
# I() allows arithmetic inside the formula
m4 = smf.ols('time_on_app ~ new_ui + account_age + I(account_age**2)', data=df).fit()
print(m4.summary())
```
